## 1. My lane as an ML task (type)

*Classification, clustering, ranking, or scoring — which one, and why?*

```python
"""Ranking/Scoring: Anyone reviewing the content from most priority to least priority with respect to which pages to refresh and revamp first. A binary classification with a yes or no doesnt work as it can flood the reviewers with large amounts of webpages with no clue as to which is prioritized, clustering is also viable for this task but still clutters the reviewers so ranking and scoring is the most optimal tasks to do."""
```

## 2. Target or proxy

*What would you predict? Where does that label come from — observed outcome or a defined rule?*

```python
""" is_declining_label is an observed outcome over time: comparing impressions_last_30d to impressions_prev_30d, where a drop of more than 20% indicates trend_direction as 'down'."""

import pandas as pd
df = pd.read_csv('../../data/raw/content_refresh_anonymized.csv')
df["is_declining_label"] = df["trend_direction"].str.lower().eq("down").astype(int)
print("Target base rate (observed decline):")
print(df["is_declining_label"].value_counts(normalize=True).round(3))
```

## 3. Success metric

*One metric you can defend. What number means 'good'?*

```python
"""
Precision@50.
A score of 0.80 or more means "good", well above the 54.2% base rate.
"""

base_rate = df["is_declining_label"].mean()
print(f"Base rate (random guessing): {base_rate:.3f} ({base_rate*100:.1f}%)")
print("Target 'good' metric: Precision@50 >= 0.800 (80%)")
```
